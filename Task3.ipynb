```python
import pandas as pd
import numpy as np

df = pd.read_csv("T3Dataset.csv")
```

```python
df.dropna(subset=['CustomerID'], inplace=True)  # drop rows where CustomerID is null
```

```python
df.dropna(subset=['Name'], inplace=True)  # drop rows where Name is null
df = df[~pd.to_numeric(df['Name'], errors='coerce').notna()] # drop rows where Name is numeric
```

```python

# Replace nulls in the Gender column with 'Unknown'
df['Gender'] = df['Gender'].fillna('Unknown')
```

```python
# Fixing the typos in the Gender column
df['Gender'] = df['Gender'].replace({
    'Femlae': 'Female',
    'mle': 'Male',
})
```

```python

# Assuming valid ages are between 0 and 100
valid_ages = df['Age'][(df['Age'].notna()) & (df['Age'] <= 100) & (df['Age'] >= 0)]

# Compute the mean of valid ages
mean_age = valid_ages.mean()

# Replace with mean age
df['Age'] = df['Age'].apply(lambda x: mean_age if (pd.isna(x) or x > 100 or x < 0) else x)
```

```python
# Round ages to the nearest integer
df['Age'] = df['Age'].round().astype(int)
```

```python
#Replace the nulls in the Country column with random countries from a predefined list
import random
countries = ['USA', 'India', 'UK', 'Canada', 'Germany']  # list of countries
df['Country'] = df['Country'].apply(lambda x: random.choice(countries) if pd.isna(x) else x)
```

```python
# Fixing the typos in the Country column

df['Country'] = df['Country'].replace({
    'Indai': 'India',
    'Canda': 'Canada',
})
```

```python
# Convert to datetime
df['SignupDate'] = pd.to_datetime(df['SignupDate'], errors='coerce')
df['LastLogin'] = pd.to_datetime(df['LastLogin'], errors='coerce')

# Drop rows with null dates
df.dropna(subset=['SignupDate', 'LastLogin'], inplace=True)

#Ensure SignupDate <= LastLogin
df = df[df['SignupDate'] <= df['LastLogin']]
```

```python
#Identify valid purchases (positive and not null)
valid_purchases = df['TotalPurchase'][(df['TotalPurchase'].notna()) & (df['TotalPurchase'] > 0)]

#Compute mean of valid purchases
mean_purchase = valid_purchases.mean()

# Replace nulls and non-positive values with mean
df['TotalPurchase'] = df['TotalPurchase'].apply(
    lambda x: mean_purchase if (pd.isna(x) or x <= 0) else x
)
```

```python
import random

# Predefined list of devices
devices = ['desktop', 'mobile']

# Replace nulls with a random choice from the list
df['PreferredDevice'] = df['PreferredDevice'].apply(
    lambda x: random.choice(devices) if pd.isna(x) else x
)
```

```python
df['PreferredDevice'] = df['PreferredDevice'].replace({
    'dasktop': 'desktop',
    'moblie': 'mobile',
})
```

```python
df = df.dropna(subset=['Email']) # drop rows where Email is null
```

```python

df = df[df["Email"].str.contains("@") & df["Email"].str.contains(".", regex=False)]
```

```python
# Save the cleaned dataset
df.to_csv("cleaned_dataset.csv", index=False)
```

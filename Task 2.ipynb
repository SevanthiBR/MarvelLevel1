```python
import pandas as pd
from faker import Faker
import random

fake = Faker()
Faker.seed(42)

data = []
for i in range(100):
    data.append({
        "Cafe_Name": fake.company(),
        "City": fake.city(),
        "Rating": round(random.uniform(3.0, 5.0), 1),
        "Price_for_Two": random.randint(200, 1200),
        "Specialty": random.choice(["Espresso", "Latte", "Cappuccino", "Mocha", "Cold Brew"]),
        "Opening_Hours": f"{random.randint(7,10)}AM - {random.randint(7,11)}PM"
    })

df = pd.DataFrame(data)
df.to_csv("T2Dataset.csv", index=False)
```

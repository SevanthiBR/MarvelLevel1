```python
import joblib
import pandas as pd

filenames = ['Model1.pkl', 'Model2.pkl', 'Model3.pkl', 'Model4.pkl', 'Model5.pkl']

for name in filenames:
    obj = joblib.load(name)
    print(f"--- {name} ---")
    print(f"Type: {type(obj)}")
    
    # Number of features
    if hasattr(obj, 'n_features_in_'):
        print(f"Features: {obj.n_features_in_}")
    
    # Feature names
    if hasattr(obj, 'feature_names_in_'):
        print(f"Feature names: {obj.feature_names_in_}")
    print("\n")
```

```python
import joblib
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score

iris = load_iris()
X_test = iris.data
y_true = iris.target

model_files = [ 'Model1.pkl', 'Model2.pkl', 'Model3.pkl', 'Model4.pkl', 'Model5.pkl']

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import joblib

results = {}

for file_name in model_files:
    
    model = joblib.load(file_name)
    
    y_pred = model.predict(X_test)
    
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average='weighted')
    rec = recall_score(y_true, y_pred, average='weighted')
    f1 = f1_score(y_true, y_pred, average='weighted')
    
    model_type = type(model).__name__
    
    print(f"{file_name:<15} | {model_type:<25} | "
          f"Acc: {acc:.4f} | Prec: {prec:.4f} | "
          f"Rec: {rec:.4f} | F1: {f1:.4f}")
    
    results[file_name] = {
        "Accuracy": acc,
        "Precision": prec,
        "Recall": rec,
        "F1": f1
    }
```
